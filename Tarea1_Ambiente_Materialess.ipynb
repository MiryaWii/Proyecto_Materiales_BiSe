2. Creación del ambiente

Posteriormente se activó mediante:

Para el estudio computacional de materiales Bi–Se:

- ASE (Atomic Simulation Environment): permite construir y manipular estructuras atómicas y preparar simulaciones.
- pymatgen: permite analizar composiciones, estructuras cristalinas y propiedades de materiales.
- spglib: permite estudiar la simetría de estructuras cristalinas.

La instalación se realizó con:

python -m pip install ase pymatgen spglib

```python
from importlib.metadata import version
from ase.data import atomic_numbers
from pymatgen.core import Composition
import spglib

# Composición del seleniuro de bismuto
material = Composition("Bi2Se3")

print("ASE:", version("ase"))
print("pymatgen:", version("pymatgen"))
print("spglib:", spglib.__version__)

print("\nMaterial:", material.reduced_formula)
print("Número atómico del Bi:", atomic_numbers["Bi"])
print("Número atómico del Se:", atomic_numbers["Se"])

print("\nPrimera prueba de materiales: OK")
```
